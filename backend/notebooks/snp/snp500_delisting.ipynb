```python
from time import sleep
import csv
from datetime import date

# Configure which year(s) to pull. Days 01-10 of March, June, September, December.
YEARS = range(2020, 2026)
MONTHS = [3, 6, 9, 12]
DAYS = range(1, 11)
OUTPUT_PATH = "snp500_delisting.csv"

rows = []
for year in YEARS:
    for month in MONTHS:
        for day in DAYS:
            d = date(year, month, day)
            try:
                records = provider.fetch_data(date=d)
            except Exception as exc:
                print(f"{d}: error -> {exc}")
                continue
            for r in records:
                rows.append({
                    "press_release_date": r.press_release_date.isoformat(),
                    "effective_date": r.effective_date,
                    "name": r.name,
                    "symbol": r.symbol,
                    "gics_sector": r.gics_sector,
                })
            print(f"{d}: {len(records)} deletion(s)")
            sleep(5.0)
        sleep(10.0)
    sleep(10.0)

fieldnames = ["press_release_date", "effective_date", "name", "symbol", "gics_sector"]
with open(OUTPUT_PATH, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(rows)

print(f"\nSaved {len(rows)} row(s) to {OUTPUT_PATH}")
```
